### Imports and settings

```python
import pandas as pd
import pyarrow.parquet as pq
from pathlib import Path

# Show every column
pd.set_option("display.max_columns", None)

PARQUET_PATH = Path("sample_data.parquet")  # change this to your file or folder
```

### File information

```python
meta = pq.ParquetFile(PARQUET_PATH).metadata

print(f"File:        {PARQUET_PATH.name} ({PARQUET_PATH.stat().st_size / 1024:,.1f} KB)")
print(f"Rows:        {meta.num_rows:,}")
print(f"Columns:     {meta.num_columns}")
print(f"Row groups:  {meta.num_row_groups}")
print(f"Created by:  {meta.created_by}")
```

### Load the data into a DataFrame

```python
df = pd.read_parquet(PARQUET_PATH, engine="pyarrow")
print(f"Loaded {df.shape[0]:,} rows x {df.shape[1]} columns")
```
